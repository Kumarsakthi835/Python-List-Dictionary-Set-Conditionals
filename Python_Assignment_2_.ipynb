```python
# --List Creation--

# Age_list with 5 integers
age_list = [24, 25, 27, 28, 29]
print("Age List:", age_list)

# Name_list with 5 strings
name_list = ["Kumar", "Priya", "Raj", "Anjali", "Vijay"]
print("Name List:", name_list)

```

```python
# Append "Yazhini" to name_list
name_list.append("Yazhini")
print("After Append:", name_list)
```

```python
# Insert 26 at index 2 in age_list
age_list.insert(2, 26)
print("After Insert:", age_list)
```

```python
#Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After Remove:", name_list)
```

```python
#Pop last element from age_list
popped = age_list.pop()
print("Popped Element:", popped)
print("After Pop:", age_list)
```

```python
# Extend age_list with [31, 30, 32]
age_list.extend([31, 30, 32])
print("After Extend:", age_list)
```

```python
#Sort age_list in descending order
age_list.sort(reverse=True)
print("After Sort Descending:", age_list)
```

```python
# max, min and sum of age_list
print("Max Age:", max(age_list))
print("Min Age:", min(age_list))
print("Sum of Ages:", sum(age_list))
```

```python
#first element of name_list
print("First Element:", name_list[0])
```

```python
#last element of name_list
print("Last Element:", name_list[-1])
```

```python
#Elements from index 2 to index 4
print("Index 2 to 4:", name_list[2:5])
```

```python
#Name_list in reverse order
print("Reverse:", name_list[::-1])
```

```python
#Student_marks dictionary
student_marks = {
    "Kumar":   85,
    "Priya":   90,
    "Raj":     75,
    "Anjali":  88,
    "Vijay":   70
}
print("Student Marks:", student_marks)
```

```python
#Access mark of a specific student
print("Priya's Mark:", student_marks["Priya"])
```

```python
# New student "Janani" with mark 80
student_marks["Janani"] = 80
print("After Adding Janani:", student_marks)
```

```python
#Update mark of existing student
student_marks["Raj"] = 82
print("After Update:", student_marks)
```

```python
# Keys, values and items
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

```python
# Set with duplicate values
my_set = {'a','e','i','o','u','a','a','i'}
print("My Set:", my_set)
```

```python
# Sets created
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set1:", set1)
print("Set2:", set2)
```

```python
#Union and Intersection of set1 and set2

#all elements from both sets
print("Union:", set1 | set2)

#common elements only
print("Intersection:", set1 & set2)
```
